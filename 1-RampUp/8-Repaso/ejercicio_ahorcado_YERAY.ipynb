## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
import random



def juego():
    listapalabras = ["casa", "botella", "mojito", "alfajor", "vaso", "chiguagua", "barco", "cine", "fabada", "baño", "cerveza"]
    palabra = random.choice(listapalabras)
    rayas = []

    for i in palabra:
        rayas.append("_")

    print(rayas)
#    print(palabra)

    final = 0
    seguir = 1 
    vidas = 6
    while seguir == 1 and final == 0:
        if vidas >0:
            letrausuario= input("Introduzca letra")


            if letrausuario in palabra:
                print("Acierto. Continua")
                for i, letra in enumerate(palabra):
                    if letrausuario == letra:
                        #print(palabra[i])
                        rayas[i] = letrausuario
                        print(rayas)
                        seguir = 1

                if "_" not in rayas:
                    final = 1
                    seguir = 0
                    print("Has ganado")
                


                

            else:
                vidas = vidas - 1
                print(f"Has fallado, te quedan {vidas} vidas")
                
        else:
            print("Fin del juego")
            final = 1
    print(palabra)
````

````python
juego()
````
