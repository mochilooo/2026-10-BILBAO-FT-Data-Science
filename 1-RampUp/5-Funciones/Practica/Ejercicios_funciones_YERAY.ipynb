## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def conversion(num):
    if num == 1:
        return "Lunes"

    elif num == 2:
        return "Martes"

    elif num == 3:
        return "Miércoles"

    elif num == 4:
        return "Jueves"

    elif num == 5:
        return "Viernes"

    elif num == 6:
        return "Sábado"

    elif num == 7:
        return "Domingo"

    else:
        return "No existe el día"


dias = int(input("Introduce el núero del día"))
print(conversion(dias))
````

````python
def semana(dia):
    semana = {1: "Lunes",
              2: "Martes",
              3: "Miércoles",
              4: "Jueves",
              5: "Viernes",
              6: "Sábado",
              7: "Domingo"}
    return semana.get(dia, "Has introducido número incorrecto")

dias = int(input("Introduce el núero del día"))
print(semana(dias))
````

## Ejercicio 2
En el ejercicio 8 de flujos de control, creábamos una pirámide invertida de números desde el 5. Crea una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide.

````python
def piramide(num):
    for n in range(num,0,-1):
        print()
        for m in range(n, 0, -1):
            print(m, end= " ")
    print()

filas = int(input("Introduce el número de filas de la pirámide"))

piramide(filas)
````

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def comp(num1, num2):
    if num1 == num2:
        return "Son iguales"

    elif num1 > num2:
        return "El primero es mayor que el segundo"

    else:
        return "El segundo es mayor que el primero"

a = int(input("Introduce el número 1"))
b = int(input("Introduce el número 2"))

print(comp(a,b))
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def contar(texto, letra):
    cont = 0
    for i in texto:
        if i == letra:
            cont = cont + 1
    return cont

a = input("Inserte el texto").lower()
b = input("Inserte la letra a buscar").lower()
print(contar(a,b))
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string, siendo la clave la letra y el valor el conteo.

````python
def diccionario(string):
    diccionario = {}

    for letra in string:
        if letra in diccionario:
            diccionario[letra] += 1
        else:
            diccionario[letra] = 1
    return diccionario

a = input("Inserte el string")

print(diccionario(a))
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

Tendrá como salida la lista

````python
def elementos(lista,comando,elemento = None):
    
    if comando == "add":
        lista.append(elemento)

    elif comando == "remove":
        lista.remove(elemento)

    else:
        print("Comano erróneo")

    print(lista)


a = [1,2,3,4,5]
b = input("Inserte el comando: (add/remove)")
c = int(input("Inserte el elemento"))

elementos(a,b,c)
````

## Ejercicio 7
Crea una función que reciba un número arbitrario de palabras, y devuelva una frase completa, separando las palabras con espacios.

````python
def frase(*palabras):
    string = ""
    for palabra in palabras:
        string = string + palabra + " "

    return string

lista = []
while True:
    a = input("Introduce Palabra o pulse 2: Finalizar")
    if(a == "2"):
        break
    else:
        lista.append(a)

frase(*lista)
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python
def fibo(num:int):
    if num <= 1:
        return num
    else:
        return fibo(num - 1) + fibo(num - 2)
    
print(fibo(int(input("Dime un numero"))))
````
