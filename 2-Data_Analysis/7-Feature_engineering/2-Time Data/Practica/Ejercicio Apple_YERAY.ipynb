### Step 5. Transform the Date column as a datetime type and check the info()

```python
fechas = pd.read_csv(url)
fechas["Date"] = pd.to_datetime(fechas["Date"])
fechas.info()
```

### Step 6.  Set the date as the index

```python
dataset = dataset.set_index("Date")
dataset.head()
```

### Step 7.  Is there any duplicate dates?

```python
dataset.index.has_duplicates
```

```python
dataset.duplicated().any()
```

### Step 8.  Ops...it seems the index is from the most recent date. Make the first entry the oldest date.

```python
dataset.sort_index(ascending=True)
dataset.head()
```

### Step 9. Get the last business day of each month

```python
dataset.index = pd.to_datetime(dataset.index) # accede al índice
```

```python
dataset.asfreq("BME") # Businnes Month End
```

### Step 10.  What is the difference in days between the first day and the oldest

```python
(dataset.index[0] - dataset.index.min()).days
```

### Step 11.  How many months in the data we have?

```python
dataset.index.to_period("M").nunique()
```

```python
len(dataset.asfreq("BME"))
```
