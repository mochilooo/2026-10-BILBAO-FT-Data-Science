### Step 1. Import the necessary libraries

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
chipo.info()
```

```python
chipo['item_price'] = pd.to_numeric(chipo['item_price'].astype(str).str.replace('$', ''), errors= 'coerce') # Cuando se produce un error produce un NaN

chipo.info()
```

```python
print(chipo.loc[chipo['item_price'].idxmax(), ['item_name', 'item_price', 'quantity']]) # Producto más caro
```

```python
chipo['unit_price'] = chipo['item_price'] / chipo['quantity'] # Producto más caro por unidad
chipo.head(10)
```

```python
print(chipo.loc[chipo['unit_price'].idxmax(), ['item_name', 'item_price', 'quantity', 'unit_price']])
```

```python
# Más de un producto
productos_caros = chipo[chipo['unit_price'] == chipo['unit_price'].max()]
print(productos_caros['item_name'].unique())
```

### Step 4. How many products cost more than $10.00?

```python
chipo[chipo['unit_price'] > 10.00][['item_name', 'item_price']]
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
chipo.loc[chipo['unit_price'] > 10]['order_id'].nunique()
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
print(chipo.groupby('order_id')['item_price'].sum())
```

```python
print((chipo.groupby('order_id')['item_price'].sum() > 10).sum())

#groupby('order_id'): agrupa las filas por pedido.
#['item_price'].sum(): suma el precio de los artículos de cada pedido.
#> 10: comprueba qué pedidos superan los 10.
#.sum(): cuenta los True, porque en Python True vale 1 y False vale 0.
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
print(chipo.loc[(chipo['item_price']>10) & (chipo['quantity']>1), 'order_id'].nunique())
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
print(chipo.loc[chipo['item_price'].idxmax(), 'quantity'])
```

```python
print(chipo.sort_values('item_price', ascending=False)['quantity'].iloc[0])
#sort_values('item_price', ascending=False): ordena las filas de mayor a menor precio.
#['quantity']: selecciona únicamente la columna de cantidad.
#.iloc[0]: obtiene el primer valor, es decir, la cantidad de la fila con el precio más alto.
#print(...): muestra el resultado.
```

```python
idx_implicito = chipo['item_price'].argmax()
idx_explicito = chipo['item_price'].idxmax()
```

```python
#usamos chipo para ver lo de los ínidces implícitos y explícitos
#chipo['item_price'].idxmax() devuelve el índice explícito
#en chipo los índices explícitos sí hacen referencia al mismo producto
print(chipo.loc[idx_explicito, 'quantity'])
```

```python
#iloc devuelve el índice implícito
print(chipo.iloc[idx_implicito]['quantity'])
```

```python
print(chipo.iloc[chipo['item_price'].argmax()]['quantity'])
```
