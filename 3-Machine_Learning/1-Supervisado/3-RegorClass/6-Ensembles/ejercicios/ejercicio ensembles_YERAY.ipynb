````python
# Nombres de las columnas
nombres = ['preg', 'plas', 'pres', 'skin', 'test', 'mass', 'pedi', 'age', 'class']

# Cargar dataset
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv'

df = pd.read_csv(url, names=nombres)

# Mostrar las primeras filas
df.head()
````

````python
from sklearn.model_selection import train_test_split

# Separar features y target
X = df.drop('class', axis=1).values
y = df['class'].values
````

### 3. Bagging
Para este apartado tendrás que crear un ensemble utilizando la técnica de bagging ([BaggingClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.BaggingClassifier.html)), mediante la cual combinarás 100 [DecisionTreeClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html). Recuerda utilizar también [cross validation](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.KFold.html) con 10 kfolds.

**Para este apartado y siguientes, no hace falta que dividas en train/test**, por hacerlo más sencillo. Simplemente divide tus datos en features y target.

Establece una semilla

````python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import KFold, cross_val_score

# Establecer semilla
seed = 42

# Crear estimador base
estimator = DecisionTreeClassifier(random_state=seed)

# Crear modelo Bagging
bag_clf = BaggingClassifier(
    estimator=estimator,
    n_estimators=100,  # Cantidad de árboles
    bootstrap=True,    # Muestreo con reemplazo
    random_state=seed
)
````

````python
# Configurar validación cruzada
kfold = KFold(
    n_splits=10,
    shuffle=True,
    random_state=seed
)

# Evaluar modelo
results = cross_val_score(
    bag_clf,
    X,
    y,
    cv=kfold,
    scoring='accuracy'
)

# Mostrar resultados
print("Accuracy (%):", results.mean() * 100)
````

### 4. Random Forest
En este caso entrena un [RandomForestClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html) con 100 árboles y un `max_features` de 3. También con validación cruzada

````python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold, cross_val_score

# Crear modelo Random Forest
rf_clf = RandomForestClassifier(
    n_estimators=100,  # Cantidad de árboles
    max_features=3,    # Variables consideradas en cada división
    random_state=42
)

# Validación cruzada: 10 K-Folds
kfold = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

# Evaluar modelo
results = cross_val_score(
    rf_clf,
    X,
    y,
    cv=kfold,
    scoring='accuracy'
)

# Mostrar resultados
print("Accuracy (%):", results.mean() * 100)
````

### 5. AdaBoost
Implementa un [AdaBoostClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.AdaBoostClassifier.html) con 30 árboles.

````python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import KFold, cross_val_score

# Crear modelo AdaBoost
ada_clf = AdaBoostClassifier(
    n_estimators=30,  # Cantidad de estimadores
    random_state=42
)

# Validación cruzada: 10 K-Folds
kfold = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

# Evaluar modelo
results = cross_val_score(
    ada_clf,
    X,
    y,
    cv=kfold,
    scoring='accuracy'
)

# Mostrar resultados
print("Accuracy (%):", results.mean() * 100)
````

### 6. GradientBoosting
Implementa un [GradientBoostingClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.GradientBoostingClassifier.html) con 100 estimadores

````python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import KFold, cross_val_score

# Crear modelo Gradient Boosting
gb_clf = GradientBoostingClassifier(
    n_estimators=100,  # Cantidad de estimadores
    random_state=42
)

# Validación cruzada: 10 K-Folds
kfold = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

# Evaluar modelo
results = cross_val_score(
    gb_clf,
    X,
    y,
    cv=kfold,
    scoring='accuracy'
)

# Mostrar resultados
print("Accuracy (%):", results.mean() * 100)
````

### 7. XGBoost
Para este apartado utiliza un [XGBoostClassifier](https://docs.getml.com/latest/api/getml.predictors.XGBoostClassifier.html) con 100 estimadores. XGBoost no forma parte de la suite de modelos de sklearn, por lo que tendrás que instalarlo con pip install

````python
from xgboost import XGBClassifier
from sklearn.model_selection import KFold, cross_val_score
````

````python
# Crear modelo XGBoost
xgb_clf = XGBClassifier(
    n_estimators=100,  # Cantidad de estimadores
    random_state=42,
    eval_metric='logloss'
)

# Validación cruzada: 10 K-Folds
kfold = KFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

# Evaluar modelo
results = cross_val_score(
    xgb_clf,
    X,
    y,
    cv=kfold,
    scoring='accuracy'
)

# Mostrar resultados
print("Accuracy (%):", results.mean() * 100)
````

### 8. Primeros resultados
Crea un dataframe con los resultados y sus algoritmos, ordenándolos de mayor a menor

````python
import pandas as pd

# Diccionario para almacenar los resultados
modelos = {
    'Bagging': bag_clf,
    'Random Forest': rf_clf,
    'AdaBoost': ada_clf,
    'Gradient Boosting': gb_clf,
    'XGBoost': xgb_clf
}

# Almacenar Accuracy medio de cada modelo
resultados = {}

for nombre, modelo in modelos.items():
    scores = cross_val_score(
        modelo,
        X,
        y,
        cv=kfold,
        scoring='accuracy'
    )

    resultados[nombre] = scores.mean()

# Crear DataFrame
df_resultados = pd.DataFrame(
    list(resultados.items()),
    columns=['Algoritmo', 'Accuracy']
)

# Ordenar de mayor a menor
df_resultados = df_resultados.sort_values(
    by='Accuracy',
    ascending=False
).reset_index(drop=True)

# Mostrar resultados
df_resultados
````

### 9. Hiperparametrización
Vuelve a entrenar los modelos de nuevo, pero esta vez dividiendo el conjunto de datos en train/test y utilizando un gridsearch para encontrar los mejores hiperparámetros.

````python
from sklearn.model_selection import train_test_split

# Separar features y target
X = df.drop('class', axis=1)
y = df['class']

# División train/test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
````

````python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score
from sklearn.ensemble import (
    BaggingClassifier,
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
````

````python
# Definir modelos
modelos = {
    'Bagging': BaggingClassifier(
        estimator=DecisionTreeClassifier(random_state=42),
        random_state=42
    ),

    'Random Forest': RandomForestClassifier(random_state=42),

    'AdaBoost': AdaBoostClassifier(
        estimator=DecisionTreeClassifier(random_state=42),
        random_state=42
    ),

    'Gradient Boosting': GradientBoostingClassifier(random_state=42),

    'XGBoost': XGBClassifier(
        random_state=42,
        eval_metric='logloss'
    )
}

# Definir hiperparámetros
param_grid = {
    'Bagging': {
        'n_estimators': [50, 100],
        'max_samples': [0.7, 1.0],
        'estimator__max_depth': [3, None]
    },

    'Random Forest': {
        'n_estimators': [50, 100, 150],
        'max_features': [2, 3, 5]
    },

    'AdaBoost': {
        'n_estimators': [30, 50, 100],
        'learning_rate': [0.1, 1.0],
        'estimator__max_depth': [1, 2]
    },

    'Gradient Boosting': {
        'n_estimators': [50, 100, 150],
        'learning_rate': [0.05, 0.1],
        'max_depth': [2, 3]
    },

    'XGBoost': {
        'n_estimators': [50, 100, 150],
        'learning_rate': [0.05, 0.1],
        'max_depth': [2, 3]
    }
}
````

````python
# Diccionario para almacenar resultados
mejores_modelos = {}

for nombre, modelo in modelos.items():

    grid = GridSearchCV(
        estimator=modelo,
        param_grid=param_grid[nombre],
        cv=5,
        scoring='accuracy',
        n_jobs=-1
    )

    # Entrenar búsqueda
    grid.fit(X_train, y_train)

    # Guardar mejor modelo
    mejores_modelos[nombre] = grid.best_estimator_

    print(f"\nModelo: {nombre}")
    print("Mejores parámetros:", grid.best_params_)
    print("Mejor Accuracy CV:", grid.best_score_)
````

````python
# Diccionario para almacenar Accuracy
resultados_grid = {}

for nombre, modelo in mejores_modelos.items():

    # Predicciones
    y_pred = modelo.predict(X_test)

    # Calcular Accuracy
    accuracy = accuracy_score(y_test, y_pred)

    resultados_grid[nombre] = accuracy

    print(f"{nombre}: {accuracy:.4f}")
````

````python
# Crear DataFrame
df_grid = pd.DataFrame(
    list(resultados_grid.items()),
    columns=['Algoritmo', 'Accuracy']
)

# Ordenar resultados
df_grid = df_grid.sort_values(
    by='Accuracy',
    ascending=False
).reset_index(drop=True)

# Mostrar resultados en porcentaje
df_grid.style.format({
    'Accuracy': '{:.2%}'
})
````
